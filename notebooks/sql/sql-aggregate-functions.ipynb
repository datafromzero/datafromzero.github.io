## Setup

Run this cell first. It builds an empty database in memory and loads the five cafe-sales tables into it.

```python
%pip install -q duckdb==2.0.0  # the version these lessons were checked with
import duckdb, urllib.request

URL = "https://raw.githubusercontent.com/datafromzero/datafromzero.github.io/v0.1.0/datasets/cafe-sales/v1/cafe-sales.sql"
script = urllib.request.urlopen(URL).read().decode("utf-8")

con = duckdb.connect()  # an empty database that lives in memory
con.execute(script)     # creates and fills the five cafe-sales tables
print("cafe-sales is loaded")
```

## Worked example

### Step 1. Count the lines

```python
con.sql("""
SELECT COUNT(*) AS order_lines FROM order_items
""").show()
```

### Step 2. Five aggregates in one query

```python
con.sql("""
SELECT COUNT(*) AS lines,
       SUM(quantity) AS units,
       ROUND(AVG(quantity), 2) AS avg_units,
       MIN(quantity) AS min_units,
       MAX(quantity) AS max_units
FROM order_items
""").show()
```

### Step 3. Three counts on one column

```python
con.sql("""
SELECT COUNT(*) AS orders,
       COUNT(customer_id) AS with_customer,
       COUNT(DISTINCT customer_id) AS customers
FROM orders
""").show()
```

### Step 4. Predict what `NULL` does

The `VALUES` list builds a table of three rows and one column named `x`, one of them `NULL`. Before you run it: will
`average` be 5 or 3.33?

```python
con.sql("""
SELECT COUNT(*) AS n,
       COUNT(x) AS non_null,
       SUM(x) AS total,
       AVG(x) AS average
FROM (VALUES (4), (6), (NULL)) AS t(x)
""").show()
```

### Step 5. An empty group

Before you run it: what will each column show when no rows are left?

```python
con.sql("""
SELECT COUNT(*) AS lines, SUM(quantity) AS units
FROM order_items
WHERE quantity > 20
""").show()
```

### Step 6. Verify against Excel and a second route

```python
con.sql("""
SELECT SUM(quantity) AS units FROM order_items WHERE order_id <> 1461
""").show()
```

```python
con.sql("""
SELECT COUNT(*) AS customers_in_table FROM customers
""").show()
```
