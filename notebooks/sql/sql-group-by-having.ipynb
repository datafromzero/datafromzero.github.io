## Setup

Run this cell first. It builds an empty database in memory and loads the five cafe-sales tables into it.

```python
%pip install -q duckdb==2.0.0  # the version these lessons were checked with
import duckdb, urllib.request

URL = "https://raw.githubusercontent.com/datafromzero/datafromzero.github.io/v0.1.0/datasets/cafe-sales/v1/cafe-sales.sql"
script = urllib.request.urlopen(URL).read().decode("utf-8")

con = duckdb.connect()  # an empty database that lives in memory
con.execute(script)     # creates and fills the five cafe-sales tables
print("cafe-sales is loaded")
```

## Worked example

### Step 1. Group by one column

Grain: one row per store.

```python
con.sql("""
SELECT store_id, COUNT(*) AS orders
FROM orders
GROUP BY store_id
ORDER BY store_id
""").show()
```

### Step 2. Group by two columns

Grain: one row per store and payment method. Before you run it: how many rows?

```python
con.sql("""
SELECT store_id, payment_method, COUNT(*) AS orders
FROM orders
GROUP BY store_id, payment_method
ORDER BY store_id, payment_method
""").show()
```

### Step 3. Filter rows before grouping with `WHERE`

Grain: one row per store, completed orders only.

```python
con.sql("""
SELECT store_id, COUNT(*) AS orders
FROM orders
WHERE status = 'completed'
GROUP BY store_id
ORDER BY store_id
""").show()
```

### Step 4. Filter groups after counting with `HAVING`

```python
con.sql("""
SELECT store_id, COUNT(*) AS orders
FROM orders
WHERE status = 'completed'
GROUP BY store_id
HAVING COUNT(*) >= 500
ORDER BY store_id
""").show()
```

### Step 5. Verify with a second route

```python
con.sql("""
SELECT COUNT(*) AS completed_orders FROM orders WHERE status = 'completed'
""").show()
```
