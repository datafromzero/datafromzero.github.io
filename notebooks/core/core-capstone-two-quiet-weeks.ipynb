## Setup

```python
# Run this cell first. Then "Run all from the top" any time: restart the session and run every cell from the first.
# Written and run with pandas 3.0.5 and 2.2.3. The only difference in the output: the dteday column is datetime64[us] in 3.0.5 and datetime64[ns] in 2.2.3.
# It downloads the Bike Sharing dataset from the UCI Machine Learning Repository (not from this site) and loads both files.
# Citation (CC BY 4.0): Fanaee-T, H. (2013). Bike Sharing [Dataset]. UCI Machine Learning Repository. https://doi.org/10.24432/C5W894
import io
import urllib.request
import zipfile

import pandas as pd

URL = "https://archive.ics.uci.edu/static/public/275/bike+sharing+dataset.zip"
raw = urllib.request.urlopen(URL, timeout=60).read()
with zipfile.ZipFile(io.BytesIO(raw)) as z:
    z.extractall("bike-sharing")

hour = pd.read_csv("bike-sharing/hour.csv", parse_dates=["dteday"])
day = pd.read_csv("bike-sharing/day.csv", parse_dates=["dteday"])
print("hour:", hour.shape, "day:", day.shape)
```
