## Setup

````python
import math
import numbers

import numpy as np
import pandas as pd

# Run this cell as it is. It also holds the answer checker used by the exercise cells.
EXERCISES = ["Exercise 1", "Exercise 2", "Exercise 3"]


RESULTS = {}


def same(got, expected):
    """True when two values agree: floats within rounding, lists and dicts item by item."""
    try:
        if isinstance(expected, dict):
            return isinstance(got, dict) and got.keys() == expected.keys() and all(same(got[k], expected[k]) for k in expected)
        if isinstance(expected, (list, tuple)):
            return isinstance(got, (list, tuple)) and len(got) == len(expected) and all(same(g, e) for g, e in zip(got, expected))
        if isinstance(expected, bool) or isinstance(got, bool):
            return type(got) is type(expected) and got == expected
        if isinstance(expected, float):
            return isinstance(got, numbers.Real) and math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
        if isinstance(expected, int):
            return isinstance(got, numbers.Integral) and got == expected
        return type(got) is type(expected) and got == expected
    except Exception:
        return False


def check(label, name, expected, traps=None):
    """Compare the variable called `name` with the known answer. It never stops Run all."""
    got = globals().get(name)
    if got is None:
        RESULTS[label] = None
        print(f"{label}: give {name} a value, then run this cell again.")
        return False
    if same(got, expected):
        RESULTS[label] = True
        print(f"{label}: correct.")
        return True
    RESULTS[label] = False
    print(f"{label}: not yet. Expected {expected!r}, but {name} is {got!r}.")
    for wrong, message in (traps or {}).items():
        if same(got, wrong):
            print("Hint:", message)
    return False
````

### Exercise 2: On your own

**Task:** How many of the 16 products cost 4.50 or more? The 16 prices are in a list for you. Before you start, say in one sentence what kind of number the answer is. Write a loop with an `if` inside it, count the matching prices and set `answer_2`.

````python
check("Exercise 2", "answer_2", 8, traps={6: "Two products cost exactly 4.50, the Latte and the Cappuccino. Does your test include 4.50?"})
````

<details>
<summary>Hint</summary>

Start a counter at 0 before the loop. Inside the loop, the `if` from the conditionals lesson decides whether to add 1.

</details>

<details>
<summary>Solution</summary>


**Task:** A colleague wants the first three prices of the list and writes the slice below, but gets only two. Fix the slice, then set `answer_3 = first_three`.

**Expected:** a list of three prices.

````python
prices = [3.00, 3.50, 4.50, 4.50]  # Espresso, Americano, Latte, Cappuccino
first_three = prices[0:2]  # the colleague expects three prices
answer_3 = None  # fix the slice above, then set answer_3 = first_three
````

````python
check("Exercise 3", "answer_3", [3.0, 3.5, 4.5], traps={(3.0, 3.5): "The end of a slice is left out: [0:2] stops before position 2. Which end gives three items?"})
````

<details>
<summary>Hint</summary>

A slice `[a:b]` includes position a and leaves out position b. Which positions do you want?

</details>

<details>
<summary>Solution</summary>

```python

```

The answer is `[3.0, 3.5, 4.5]`. `[0:2]` stops before position 2, so it holds positions 0 and 1 only.
