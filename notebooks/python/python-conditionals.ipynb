## Setup

````python
import math
import numbers

import numpy as np
import pandas as pd

# Run this cell as it is. It also holds the answer checker used by the exercise cells.
EXERCISES = ["Exercise 1", "Exercise 2", "Exercise 3"]


RESULTS = {}


def same(got, expected):
    """True when two values agree: floats within rounding, lists and dicts item by item."""
    try:
        if isinstance(expected, dict):
            return isinstance(got, dict) and got.keys() == expected.keys() and all(same(got[k], expected[k]) for k in expected)
        if isinstance(expected, (list, tuple)):
            return isinstance(got, (list, tuple)) and len(got) == len(expected) and all(same(g, e) for g, e in zip(got, expected))
        if isinstance(expected, bool) or isinstance(got, bool):
            return type(got) is type(expected) and got == expected
        if isinstance(expected, float):
            return isinstance(got, numbers.Real) and math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
        if isinstance(expected, int):
            return isinstance(got, numbers.Integral) and got == expected
        return type(got) is type(expected) and got == expected
    except Exception:
        return False


def check(label, name, expected, traps=None):
    """Compare the variable called `name` with the known answer. It never stops Run all."""
    got = globals().get(name)
    if got is None:
        RESULTS[label] = None
        print(f"{label}: give {name} a value, then run this cell again.")
        return False
    if same(got, expected):
        RESULTS[label] = True
        print(f"{label}: correct.")
        return True
    RESULTS[label] = False
    print(f"{label}: not yet. Expected {expected!r}, but {name} is {got!r}.")
    for wrong, message in (traps or {}).items():
        if same(got, wrong):
            print("Hint:", message)
    return False

START = "# --- your code starts here ---"


def rerun_cell(label, marker, var, cases):
    """Run the code under START in your own cell again with other inputs. Each case is (inputs, expected)."""
    mine = [cell for cell in In if marker in cell and "rerun_cell(" not in cell]
    if not mine or START not in mine[-1]:
        RESULTS[label] = None
        print(f"{label}: run your exercise cell first (keep the START line in it), then run this cell again.")
        return False
    body = mine[-1].split(START, 1)[1]
    for inputs, expected in cases:
        names = dict(inputs)
        try:
            exec(body, names)
        except Exception as error:
            RESULTS[label] = False
            print(f"{label}: not yet. With {inputs} your code stopped with {type(error).__name__}: {error}")
            return False
        if not same(names.get(var), expected):
            RESULTS[label] = False
            print(f"{label}: not yet. With {inputs}, {var} should be {expected!r}, but your code gave {names.get(var)!r}.")
            return False
    RESULTS[label] = True
    print(f"{label}: correct ({len(cases)} cases).")
    return True
````

## Unit 2: if, elif and else

**Predict first.** A Turkey Sandwich costs 9.25, which is at least 8 and also at least 4.5. Which tier does the chain choose?

````python
price = 9.25
if price >= 8:
    tier = "premium"
elif price >= 4.5:
    tier = "regular"
else:
    tier = "cheap"
tier
````

## Unit 4: Flatten a nested condition

**Predict first.** An `if` inside an `if`. Which tier does it give for 4.50?

````python
price = 4.5
if price >= 4.5:
    if price >= 8:
        tier = "premium"
    else:
        tier = "regular"
else:
    tier = "cheap"
tier
````
