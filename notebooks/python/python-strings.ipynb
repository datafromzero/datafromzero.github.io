## Setup

````python
import math
import numbers

import numpy as np
import pandas as pd

# Run this cell as it is. It also holds the answer checker used by the exercise cells.
EXERCISES = ["Exercise 1", "Exercise 2", "Exercise 3"]


RESULTS = {}


def same(got, expected):
    """True when two values agree: floats within rounding, lists and dicts item by item."""
    try:
        if isinstance(expected, dict):
            return isinstance(got, dict) and got.keys() == expected.keys() and all(same(got[k], expected[k]) for k in expected)
        if isinstance(expected, (list, tuple)):
            return isinstance(got, (list, tuple)) and len(got) == len(expected) and all(same(g, e) for g, e in zip(got, expected))
        if isinstance(expected, bool) or isinstance(got, bool):
            return type(got) is type(expected) and got == expected
        if isinstance(expected, float):
            return isinstance(got, numbers.Real) and math.isclose(got, expected, rel_tol=1e-9, abs_tol=1e-9)
        if isinstance(expected, int):
            return isinstance(got, numbers.Integral) and got == expected
        return type(got) is type(expected) and got == expected
    except Exception:
        return False


def check(label, name, expected, traps=None):
    """Compare the variable called `name` with the known answer. It never stops Run all."""
    got = globals().get(name)
    if got is None:
        RESULTS[label] = None
        print(f"{label}: give {name} a value, then run this cell again.")
        return False
    if same(got, expected):
        RESULTS[label] = True
        print(f"{label}: correct.")
        return True
    RESULTS[label] = False
    print(f"{label}: not yet. Expected {expected!r}, but {name} is {got!r}.")
    for wrong, message in (traps or {}).items():
        if same(got, wrong):
            print("Hint:", message)
    return False
````

**Task:** Build the receipt line for 2 Lattes at 4.50 dollars each: `Latte x 2 = $9.00`. Use an f-string, calculate the amount inside it with `quantity * price`, and show two decimals. Before you start, say in one sentence how the line will look. Set `answer_2` to the text.

**Expected:** one piece of text.

<details>

An f-string can hold a calculation inside the braces. The format code goes after a colon, inside the same braces.

## Check yourself

### Exercise 3: Check yourself: spot the flaw

**Task:** A colleague wants the product name in capitals and writes the code below, but the label still shows `Croissant`. Fix the code so that `label` holds the capitals, then set `answer_3` to `label`.

**Expected:** one piece of text, in capital letters.

````python
product = "Croissant"
product.upper()
label = product  # the colleague expects CROISSANT here
answer_3 = None  # fix the lines above so that label holds the capitals, then set answer_3 = label
````

````python
check("Exercise 3", "answer_3", "CROISSANT", traps={"Croissant": "upper() gave back a new string and the code threw it away. Keep it: label = product.upper()."})
````

<details>
<summary>Hint</summary>

A string method does not change the string it is called on. Where does the new string go in the colleague's code?

</details>

<details>
<summary>Solution</summary>

```python
label = product.upper()
```

The answer is `'CROISSANT'`. `product.upper()` on a line of its own gives back a new string that nothing keeps.

</details>
